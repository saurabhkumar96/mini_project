Pandas Python ki ek library hai jo mainly data ko read, clean, analyze aur manipulate karne ke liye use hoti hai.
Example: maan lo students ka data hai:

Name	Age	Marks
Rahul	20	85
Priya	21	92
Amit	19	76


Pandas se aap easily:
    CSV/Excel files read kar sakte ho
    rows/columns select kar sakte ho
    missing data handle kar sakte ho
    filtering kar sakte ho
    sorting kar sakte ho
    average, sum, count nikal sakte ho
    data ko group karke analyze kar sakte ho

```python
import pandas as pd
```

```python

data = {
    "Name": ["Rahul", "Priya", "Amit"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 76]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print(df.shape)
```

```python
print(df.columns)
print(df.info())
print(df.describe())
```

Column select karna

```python
print(df["Name"])
print(df[["Name","Marks"]])
```

Filtering — bahut important

```python
print(df[df["Marks"] > 80])
print(df[(df["Marks"] > 80) & (df["Age"] > 20)])
```
