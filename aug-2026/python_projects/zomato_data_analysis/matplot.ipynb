```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.title("my-first Graph")
plt.xlabel("x-value")
plt.ylabel("y-value")
```

```python
plt.plot(x,y,color="red",linestyle="--",marker="*")
plt.show()
```

```python
names = ["A", "B", "C", "D"]
marks = [80, 65, 90, 75]

plt.bar(names,marks)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 3, 6, 2, 7]

plt.scatter(x,y)
```

```python
subjects = ["Math", "Science", "English", "Hindi"]
marks = [30, 25, 20, 25]
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
```

```python
marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]
plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```

## Aapko kis order mein seekhna chahiye?
plot() — Line graph
title(), xlabel(), ylabel()
Colors, markers aur line styles
bar() — Bar chart
scatter() — Scatter plot
pie() — Pie chart
hist() — Histogram
Multiple lines
subplot() — Multiple graphs
Legends, grid aur styling
Pandas DataFrame ke saath Matplotlib
Real-world datasets par visualization
